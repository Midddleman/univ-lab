```python
import pandas as pd

# ---------- 1. 读取 CSV ----------
df_yield = pd.read_csv("../../CSV/IAMC/Basin_yield_filtered.csv")
df_area  = pd.read_csv("../../CSV/gridarea/Basin_area.csv")

# ---------- 2. 只保留 CEREAL ----------
df_yield = df_yield[df_yield["land"] == "CEREAL"].copy()


# ---------- 3. 构造 Country_Basin（与 yield 对齐） ----------
df_area["basin_full"] = df_area["country"] + "_" + df_area["basin"]
df_yield["country"] = df_yield["basin"].str.split("_").str[0]
df_yield["17region"] = df_yield["country"].map(mapping)
df_yield["basin_full"]=df_yield["basin"]
```

```python
df = df_yield.merge(
    df_area[["basin_full", "Value"]],
    on=["basin_full"],
    how="left")

df = df.drop(columns=["country", "basin_full"])
df = df.rename(columns={
    "Value_x": "yield",
    "Value_y": "area"
})
```

```python
df["prod"] = df["yield"] * df["area"]

# ---------- 8. basin → 17 region（面积加权） ----------
df_region = (
    df.groupby(["17region", "year"])
      .agg(
          prod_sum=("prod", "sum"),
          area_sum=("area", "sum")
      )
      .reset_index()
)

df_region["yield_region"] = df_region["prod_sum"] / df_region["area_sum"]

df_region = df_region[["17region", "year", "yield_region"]]
```

```python
# ---------- 9. 保存 ----------
df_region.to_csv("../../CSV/IAMC/Basin_yield_17regions.csv", index=False)
```
