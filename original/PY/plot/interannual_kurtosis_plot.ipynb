```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("../../CSV/interannual_diff_kurtosis.csv")

plt.figure(figsize=(8, 6))
plt.plot(df["year"], df["kurt_region"], label="Region mode",color='orange',marker='o')
plt.plot(df["year"], df["kurt_basin"], label="Basin mode",color='blue',marker='o')

plt.xlabel("Year", fontsize=14)
plt.ylabel("Kurtosis", fontsize=14)
plt.legend(loc="lower right", fontsize=12)
plt.grid(True)
plt.show()
plt.savefig("../../plot/interannual_diff_kurtosis.png", dpi=300)
```
